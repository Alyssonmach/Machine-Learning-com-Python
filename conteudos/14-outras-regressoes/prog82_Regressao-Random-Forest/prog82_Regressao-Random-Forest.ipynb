# Regressão com Random Forest

O Random Forest também pode ser usado em tarefas de regressão.

Anteriormente, vimos como usar árvores de decisão em problemas de regressão. O **Random Forest** combina várias árvores para produzir previsões. Essa abordagem é chamada de **aprendizado em conjunto** (*ensemble learning*).
