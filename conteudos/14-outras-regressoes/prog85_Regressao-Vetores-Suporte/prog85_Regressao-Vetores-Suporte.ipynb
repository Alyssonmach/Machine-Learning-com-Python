# Regressão com vetores de suporte

Uma introdução à regressão com máquinas de vetores de suporte (SVR).

- O SVR compartilha conceitos com o SVM usado em classificação.
- A tarefa de regressão estima valores numéricos.
- O parâmetro **epsilon** define uma margem em torno das previsões; erros dentro dessa margem podem ser tratados de forma diferente pelo modelo.
