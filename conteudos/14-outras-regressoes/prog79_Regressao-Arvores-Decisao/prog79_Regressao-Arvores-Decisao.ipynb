# Regressão com árvores de decisão

As árvores de decisão também podem ser usadas para estimar valores numéricos.

A regressão com árvores de decisão segue uma lógica semelhante à classificação. Para rever o funcionamento das árvores, consulte os materiais sobre classificação. Neste caso, as folhas do modelo representam valores numéricos.

O modelo pode ser ajustado com parâmetros específicos para tarefas de regressão.
