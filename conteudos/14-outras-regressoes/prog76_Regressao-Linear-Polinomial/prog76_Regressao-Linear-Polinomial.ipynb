# Regressão polinomial

Vamos conhecer uma extensão da regressão linear que pode modelar relações não lineares.

A regressão linear simples pode não representar bem relações não lineares. A **regressão polinomial** transforma as variáveis de entrada e ajusta um modelo linear a essas transformações. Assim, pode modelar relações mais complexas.

O modelo polinomial pode ser representado pela seguinte equação:

$y = b_{0} + b_{1} * x + b_{2} * x^{2} + .. + b_{n} * x^{n}$
