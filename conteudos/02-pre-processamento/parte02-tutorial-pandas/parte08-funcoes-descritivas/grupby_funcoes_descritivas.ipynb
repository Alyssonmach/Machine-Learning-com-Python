# Usando `groupby` e funções descritivas

Neste material, vamos selecionar e agrupar dados com `groupby` e aplicar algumas funções descritivas. Assim, ampliaremos nosso conhecimento de pré-processamento com Python.

```python
# importando o pacote Pandas do Python
import pandas as pd
```

```python
# carregando os dados do dataset em um objeto
dados = pd.read_csv("bank.csv", encoding = "UTF-8", sep = ";")
```

```python
# visualizando os primeiros exemplos treináveis
dados.head()
```

O método `describe` resume estatísticas do conjunto de dados. Por padrão, essas estatísticas são calculadas para as colunas numéricas.

```python
# fazendo uma análise estatística nos dados
dados.describe()
```

A seguir, vamos explicar o que cada função descritiva representa e como ela pode ajudar a identificar dados inconsistentes ou ausentes.

Por exemplo, `mean` calcula a média de uma coluna numérica. Também podemos usar essa função diretamente para obter a média de uma coluna específica. Veja o exemplo:

```python
# funções de agrupamento - média das idades
dados.age.mean()
```

O resultado corresponde ao valor de `mean` exibido por `describe`.

Também podemos calcular as demais estatísticas descritivas.

```python
# funções de agrupamento - valor mínimo das idades
dados.age.min()
```

```python
# funções de agrupamento - valor máximo das idades
dados.age.max()
```

Agora, vamos calcular estatísticas para um subconjunto de usuários. No exemplo, veremos a média de idade das pessoas casadas.

```python
# média de idade das pessoas casadas
dados[dados.marital == "married"].age.mean()
```

Agora, vamos usar `groupby` para agrupar os dados por estado civil e escolaridade e calcular a média de idade em cada grupo.

```python
# usando o atributo "groupby" para fazer uma seleção mais precisa
dados[["marital", "education", "age"]].groupby(["marital", "education"]).mean()
```
