# Usando o método `isin`

Neste material, vamos usar o método `isin` para filtrar dados. Ele permite selecionar registros que correspondem a um conjunto de valores e, assim, refinar o pré-processamento em Python.

```python
# importando o pacote Pandas do Python
import pandas as pd
```

```python
# carregando os dados do dataset em um objeto
dados = pd.read_csv("bank.csv", encoding = "UTF-8", sep = ";")
```

```python
# visualizando os primeiros exemplos treináveis do dataset
dados.head()
```

```python
# dimensão dos dados
dados.shape
```

Vamos filtrar os dados para selecionar pessoas casadas que tenham concluído o ensino fundamental ou o ensino médio.
- casada e (ensino fundamental ou ensino médio);

```python
# dimensão dos dados filtrados
dados[(dados.marital == "married") & ((dados.education == "primary") | 
                                      (dados.education == "secondary"))].shape
```

Também podemos escrever esse filtro de outra forma e obter o mesmo resultado:

```python
# dimensão dos dados filtrados
dados[(dados.marital == "married") & (dados.education.isin(["primary", "secondary"]))].shape
```

Agora, vamos usar o método `isin` da biblioteca **Pandas** para fazer essa seleção.

```python
# criando um dicionario para filtrar mais os dados
filtro = dados.isin({"marital" : ["married"], "education" : ["primary", "secondary"]})
# dimensão dos dados filtrados
dados[(filtro.marital) & (filtro.education)].shape
```

Assim, usamos `isin`, da biblioteca **Pandas**, para filtrar os dados de acordo com uma lista de valores.
