# Tipos de Variáveis

Em uma base de dados tabular, cada linha costuma representar um exemplo, e cada coluna, uma variável. É importante conhecer os tipos dessas variáveis para escolher como pré-processar os dados e evitar erros no algoritmo de aprendizado de máquina. Uma base bem organizada facilita a análise e contribui para a qualidade dos resultados.

### Tipos de variáveis numéricas

- **Contínuas**: podem assumir qualquer valor em um intervalo, como números reais.
- **Discretas**: assumem valores contáveis, geralmente inteiros.

### Tipos de variáveis categóricas (texto)

- **Nominais**: categorias sem uma ordem natural.
- **Ordinais**: categorias que podem ser organizadas em uma ordem.
