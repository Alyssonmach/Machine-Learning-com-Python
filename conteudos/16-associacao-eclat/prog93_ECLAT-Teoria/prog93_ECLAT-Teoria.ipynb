# Algoritmo ECLAT

Uma introdução ao algoritmo ECLAT e à mineração de conjuntos frequentes.

- O ECLAT busca conjuntos frequentes de itens.
- O Apriori pode gerar regras de associação, que podem ser avaliadas por medidas como confiança e lift.
- O ECLAT usa o suporte para identificar conjuntos frequentes.

Etapas gerais:
1. Definir o suporte mínimo.
2. Encontrar os conjuntos de itens que atendem a esse limite.
3. Organizar os conjuntos de acordo com o suporte.
