# Combinação de classificadores

Combinar modelos pode reunir as previsões de diferentes classificadores.

### Vantagens de combinar classificadores

Podemos combinar classificadores baseados em algoritmos diferentes. Nos materiais anteriores, comparamos vários modelos em bases de crédito e do censo. Aqui, vamos combinar previsões para reduzir a dependência de um único classificador. Entre as estratégias possíveis estão:

- Combinar previsões pode melhorar o desempenho, dependendo dos dados e dos modelos escolhidos.

- A combinação permite avaliar diferentes conjuntos de classificadores.

- A combinação pode reduzir o impacto de escolher um modelo individual inadequado.

### Métodos principais

As previsões podem ser combinadas de diferentes formas. Três estratégias comuns são:

1. **Unanimidade**: todos os classificadores concordam com a mesma previsão.
2. **Maioria**: mais da metade dos classificadores escolhe a mesma classe.
3. **Pluralidade**: escolhe-se a classe com mais votos, mesmo que não tenha mais da metade.
