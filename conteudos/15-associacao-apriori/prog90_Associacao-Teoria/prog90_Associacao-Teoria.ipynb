# Introdução às regras de associação

Vamos estudar como as regras de associação ajudam a encontrar padrões em dados.

### O que são regras de associação?

![regras-associacao-ex](Imagens/regras-associacao-ex.jpeg)

As **regras de associação** identificam relações entre itens que aparecem juntos com frequência. Elas podem ajudar a analisar cestas de compras, organizar produtos, planejar promoções e criar sistemas de recomendação.

Veja alguns exemplos de aplicações:

- Organização de produtos nas prateleiras de um mercado;
- Promoções de produtos comprados em conjunto;
- Planejamento de catálogos e folhetos promocionais;
- Análise de fatores associados à evasão universitária;
- Sistemas de recomendação;
- Análise de cestas de compras (*market basket analysis*).

Padrões que são difíceis de perceber manualmente podem ser encontrados por algoritmos quando há dados adequados. Um exemplo é o **Apriori**, usado para identificar conjuntos frequentes de itens.

### Algoritmo Apriori

O algoritmo Apriori pode ser descrito em duas etapas principais:

#### 1ª etapa: encontrar conjuntos frequentes

Identificamos os conjuntos de itens cujo **suporte** é igual ou superior ao limite definido.

O suporte é calculado da seguinte forma:

**Suporte** = número de registros que contêm os itens do conjunto / número total de registros.

O suporte mede a frequência com que um item ou conjunto de itens aparece na base de dados. Primeiro, selecionamos os itens que atingem o suporte mínimo. Em seguida, formamos conjuntos maiores e calculamos novamente suas frequências. O processo continua até que não haja conjuntos maiores que atendam ao limite estabelecido.

#### 2ª etapa: gerar regras de associação

A partir dos conjuntos frequentes, geramos regras cujo nível de **confiança** seja igual ou superior ao limite definido.

A confiança é calculada da seguinte forma:

**Confiança** = número de registros que contêm X e Y / número de registros que contêm X.

A confiança mede a frequência com que Y aparece nos registros que contêm X. Outra medida usada para avaliar uma regra é o **lift**, que compara a frequência observada de X e Y com a frequência esperada caso fossem independentes.
