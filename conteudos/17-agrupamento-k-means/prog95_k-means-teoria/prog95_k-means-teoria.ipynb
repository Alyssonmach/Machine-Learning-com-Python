# Algoritmo K-means

Uma introdução ao método de agrupamento K-means.

### Exemplos de agrupamento (*clusters*)

Os algoritmos de agrupamento organizam dados sem rótulos em **k** grupos, de acordo com a semelhança entre os exemplos.

Algumas aplicações de agrupamento incluem:

- Segmentação de mercado;
- Identificação de grupos de clientes com interesses semelhantes;
- Agrupamento de documentos e notícias;
- Identificação de produtos similares;
- Segmentação de perfis de usuários;
- Análise de redes sociais.

### Características dos métodos de agrupamento

1. São métodos de aprendizado **não supervisionado**.
2. Organizam os dados em grupos de acordo com uma medida de semelhança.
3. O número de grupos **k** pode ser definido previamente, como no K-means.
4. Os grupos não têm rótulos conhecidos durante o treinamento.
5. A interpretação dos grupos depende do problema e dos dados.
6. Em geral, exemplos do mesmo grupo são mais semelhantes entre si do que a exemplos de outros grupos.

### Algoritmo de Lloyd (K-means)

![k-means-ex](Imagens/k-means-ex.png)

O algoritmo de Lloyd, usado pelo K-means, segue estas etapas:

1. Inicializar os centroides.
2. Atribuir cada ponto ao centroide mais próximo, de acordo com uma medida de distância.
3. Recalcular cada centroide como a média dos pontos atribuídos a ele.
4. Repetir as etapas 2 e 3 até que as atribuições ou os centroides parem de mudar significativamente.

### K-means++

O **K-means++** é uma estratégia de inicialização dos centroides que pode reduzir a chance de começar com posições inadequadas.

- O primeiro centroide é escolhido aleatoriamente.
- Os centroides seguintes são escolhidos com maior probabilidade quando estão distantes dos centroides já selecionados.

### Como escolher o número de grupos

Não há uma regra que determine o melhor número de grupos para todos os conjuntos de dados. Algumas estratégias incluem:

1. Usar conhecimento prévio sobre o problema para definir uma quantidade inicial de grupos.

2. Quando não há conhecimento prévio, usar uma regra aproximada como ponto de partida. O resultado ainda precisa ser avaliado no contexto dos dados.

> $clusters = \sqrt{\frac{N}{2}}$

3. Testar diferentes valores de **k** e comparar os resultados com o método do cotovelo (*elbow method*).

![elbow-method](Imagens/elbow-method.png)

No gráfico, a redução da distância se torna menor após **k = 4**. Esse ponto de inflexão pode ser usado como uma estimativa inicial do número de grupos.
