# Algoritmo KNN

Introdução ao algoritmo KNN para classificação.

O **KNN** (*k-nearest neighbors*, ou k vizinhos mais próximos) é um algoritmo simples usado em inteligência artificial. Pode ser aplicado a problemas de classificação e de regressão. Na classificação, prevê uma classe; na regressão, estima um valor numérico com base nas características do exemplo.

O KNN é chamado de método **preguiçoso** (*lazy learning*) porque não constrói um modelo explícito durante o treinamento. Ele armazena os exemplos e calcula as distâncias quando precisa fazer uma previsão. Isso reduz o custo do treinamento, mas pode tornar as previsões mais lentas.

O KNN compara os exemplos usando uma medida de distância, como a **distância euclidiana**.

- Cálculo da Distância Euclidiana (DE):
> $DE(x,y)=\sqrt[2]{\sum_{i}^{p}(x_{1}-y_{1})^{2}}$

> Quanto maior a distância euclidiana entre dois pontos, mais diferentes eles são segundo as variáveis consideradas. Quanto menor a distância, mais próximos eles estão.

Como essa ideia funciona na prática?

##### Exemplo

|**Filmes**|**Violência**|**Romance**|**Ação**|**Comédia**|
|:---------|:------------|:----------|:-------|-----------|
|Invocação do Mal|0.6|0.0|0.3|0.0|
|Floresta Maldita|0.9|0.0|0.5|0.1|
|Meu Passado me Condena|0.1|0.2|0.1|0.9|
|Vizinhos 2|0.0|0.1|0.2|0.8|
|Deadpool|0.5|0.0|0.7|0.1|
|Tirando o Atraso|0.0|0.2|0.2|0.8|

A tabela apresenta seis filmes e valores que representam características de seus gêneros. Com dados desse tipo, podemos criar um sistema de recomendação semelhante aos usados por serviços de streaming.

Para recomendar filmes, podemos comparar as características de um filme assistido com as dos demais. Distâncias menores indicam maior semelhança; distâncias maiores, menor semelhança.

Vamos comparar *Invocação do Mal* com *Floresta Maldita* e *Tirando o Atraso*:

$DE(Invocação do Mal, Floresta Maldita)=\sqrt[2]{(0.6 -0.9)^{2} + (0.0-0.0)^{2} +(0.3 - 0.5) + (0.0 - 0.1)} = 0.37$

$DE(Invocação do Mal, Tirando o Atraso)=\sqrt[2]{(0.6 -0.0)^{2} + (0.0-0.2)^{2} +(0.3 - 0.2) + (0.0 - 0.8)} = 1.02$

A distância entre *Invocação do Mal* e *Floresta Maldita* é menor do que a distância entre *Invocação do Mal* e *Tirando o Atraso*. Segundo essas características, os dois primeiros filmes são mais semelhantes.

O mesmo cálculo pode ser feito para todos os filmes da base. Em geral, dados mais completos e confiáveis podem contribuir para recomendações melhores.
