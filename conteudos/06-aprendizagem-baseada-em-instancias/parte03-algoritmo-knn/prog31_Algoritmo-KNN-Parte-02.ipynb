# Algoritmo KNN

O **KNN** (*k-nearest neighbors*, ou k vizinhos mais próximos) prevê a classe de um registro com base nas classes dos exemplos mais próximos. Em geral, quanto mais vizinhos de uma classe houver entre os exemplos próximos, maior será a chance de o registro receber essa classe.

O parâmetro **K** define quantos vizinhos serão considerados na previsão. Em uma tarefa de classificação, a classe mais frequente entre esses vizinhos costuma ser escolhida.

Muitos métodos constroem um modelo durante o treinamento. Métodos baseados em instâncias, como o KNN, armazenam os exemplos de treinamento e os consultam para calcular distâncias quando recebem um novo registro.

Por isso, a generalização ocorre quando uma nova instância precisa ser classificada; esse tipo de método também é chamado de **aprendizado preguiçoso** (*lazy learning*).

#### Fórmula da distância euclidiana

$DE(X,Y) = \sqrt{\sum^{p}_{i}(x_{i} - y_{i})^{2}}$

A distância euclidiana foi apresentada no exemplo de recomendação de filmes do material anterior.

#### KNN: variáveis na mesma escala

Ao usar o KNN, é importante colocar as variáveis numéricas em escalas comparáveis. Caso contrário, uma variável com valores muito altos, como renda anual, pode dominar o cálculo em relação a outra, como idade ou anos de estudo. O escalonamento ajuda a evitar esse efeito.

#### Algumas considerações sobre o KNN

- É um algoritmo simples que pode ser eficaz.
- Pode representar relações complexas entre as características.
- Um valor pequeno de **K** torna o método mais sensível a ruídos e outliers.
- Um valor grande de **K** pode favorecer a classe mais frequente.
- As previsões podem ser lentas, especialmente em bases grandes.

#### Outras medidas de distância

Além da distância euclidiana, existem outras medidas que podem ser usadas, conforme o tipo de dado e o problema. Alguns exemplos são:
- correlação de Pearson;
- índice de Tanimoto;
- distância Manhattan (*city block*).
