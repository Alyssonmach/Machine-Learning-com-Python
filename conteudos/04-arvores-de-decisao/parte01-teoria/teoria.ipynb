# Árvores de decisão: teoria

Árvores de decisão são métodos supervisionados e não paramétricos usados em tarefas de classificação e regressão.

Neste material, vamos estudar um algoritmo bastante usado em aprendizado de máquina: **árvores de decisão**. Esse método é supervisionado e pode ser aplicado tanto à classificação quanto à regressão. A sigla CART vem do inglês *Classification and Regression Trees*.

#### Fundamentos das árvores de decisão

Uma árvore de decisão representa uma sequência de decisões em forma de árvore. O algoritmo começa pela raiz e divide os dados em ramos, de acordo com os atributos preditores. As folhas representam as classes previstas ou os valores estimados.

Em computação, uma árvore é uma estrutura de dados formada por nós. A raiz é o nó inicial; os demais nós se organizam em níveis e podem ter nós filhos. Em uma árvore de decisão, os nós representam regras ou decisões.

> "*Árvores, de um modo geral, são estruturas de dados formadas por um conjunto de elementos que armazenam informações chamadas nós. Além disso, toda árvore possui um nó chamado raiz, que possui o maior nível hierárquico (o ponto de partida) e ligações denominadas filhos. Esses filhos podem possuir seus própios filhos que por sua vez também possuem os seus. Uma árvore de decisão nada mais é que uma árvore que armazena regras nos seus nós, e eles representam decisões a serem tomadas*".

O algoritmo é supervisionado porque aprende a partir de exemplos que incluem os resultados esperados. É não paramétrico porque não pressupõe uma forma específica para a distribuição dos dados.

Durante o treinamento, o algoritmo constrói uma árvore a partir dos registros disponíveis. Também separamos os dados para treinamento e teste e distinguimos os atributos preditores da classe-alvo.

#### Hierarquia das decisões

O algoritmo organiza as decisões em uma hierarquia: os primeiros atributos escolhidos ficam próximos à raiz, e os demais aparecem nos níveis seguintes. As folhas contêm a classe prevista ou o resultado da regressão; os nós intermediários representam decisões baseadas nos atributos preditores.

Duas medidas usadas para escolher atributos são a entropia e o ganho de informação:

- $Entropia(S) = \underbrace{ \sum_{i=1}^{c} -pi.log_{2}.pi } \\ $  

> **pᵢ** = proporção de exemplos pertencentes à classe *i*

- $Ganho(S,A) = Entropia(S) - \sum_{Values(A)}\frac{|S_{v}|}{S}Entropia(S_{v})$

O ganho de informação usa a entropia para avaliar qual atributo separa melhor os dados. O algoritmo considera os possíveis valores do atributo e calcula a redução de entropia em cada divisão. Em seguida, repete o processo nos nós filhos, usando os registros que chegaram a cada nó.

Não é necessário implementar esses cálculos manualmente para usar árvores de decisão: bibliotecas como **scikit-learn** já oferecem essa funcionalidade. Ainda assim, entender a lógica ajuda a interpretar o modelo e suas decisões.

#### Poda da árvore

Depois de construir a árvore, podemos podá-la, removendo ramos que não ajudam a generalizar ou que podem prejudicar o modelo. A poda controla a complexidade da árvore e pode reduzir o sobreajuste (*overfitting*).

Dois conceitos ajudam a entender o comportamento do modelo:

- **Viés (*bias*)**
> Erros associados às simplificações e suposições do modelo.

- **Variância**
> Sensibilidade do modelo a pequenas mudanças nos dados. Uma variância alta pode levar ao sobreajuste (*overfitting*): o modelo se ajusta demais aos dados de treinamento e tem dificuldade para generalizar.

A poda pode melhorar a capacidade de generalização, mas, se for excessiva, pode reduzir a capacidade preditiva do modelo. Avalie o desempenho em dados de validação após cada ajuste e mantenha as alterações que melhoram os resultados.

#### Vantagens das árvores de decisão

- São fáceis de interpretar.
- Não exigem padronização ou normalização das variáveis numéricas.

#### Desvantagens das árvores de decisão

- Podem gerar árvores muito complexas.
- Pequenas mudanças nos dados podem produzir uma árvore diferente.
- Encontrar a árvore ótima pode ser computacionalmente difícil.

#### Curiosidade

Árvores de decisão tiveram bastante destaque nos anos 1990. O desenvolvimento de métodos como **Random Forest** ampliou suas aplicações: essa técnica combina várias árvores e pode ser usada em classificação e regressão. A Microsoft, por exemplo, utilizou métodos de visão computacional e aprendizado de máquina em recursos associados ao Kinect, no console Xbox.
