# Random Forest

O método combina várias árvores de decisão para produzir previsões.

Neste material, vamos estudar o **Random Forest**, método que combina árvores de decisão. Essa abordagem pode ser aplicada tanto a problemas de classificação quanto de regressão.

#### Como funciona o Random Forest

O **Random Forest** combina as previsões de várias árvores de decisão. Em problemas de classificação, a classe final costuma ser escolhida por votação; em problemas de regressão, as previsões podem ser combinadas, por exemplo, pela média. Essa combinação é chamada de **aprendizado em conjunto** (*ensemble learning*).

![Exemplo de Random Forest](random-forest.png)

A combinação de vários modelos pode melhorar a estabilidade das previsões. Na classificação, as árvores votam nas classes; na regressão, suas previsões podem ser agregadas pela média.

Árvores muito complexas podem se ajustar demais aos dados de treinamento, causando **sobreajuste** (*overfitting*). Por isso, é importante avaliar o modelo em dados que não foram usados no treinamento.

Cada árvore é treinada com uma amostra dos dados e considera um subconjunto de atributos. Essa aleatoriedade ajuda a reduzir a semelhança entre as árvores. A decisão final combina as previsões de todas elas. Critérios como a impureza de Gini podem ser usados para avaliar a qualidade das divisões em cada árvore.
