# Rede neural artificial

Uma introdução ao funcionamento das redes neurais artificiais.

### Introdução

Em 1943, **Warren McCulloch** e **Walter Pitts** apresentaram um modelo lógico inspirado no neurônio biológico. Em 1958, **Frank Rosenblatt** desenvolveu o **Perceptron**, um modelo de neurônio artificial que impulsionou o interesse pelas redes neurais. Desde então, pesquisadores vêm aprimorando esses modelos e ampliando suas aplicações.

A inteligência computacional desenvolve métodos inspirados, em parte, nas capacidades dos sistemas biológicos. A combinação de métodos também pode ser útil para resolver determinados problemas.

### O que é uma rede neural?

Uma **rede neural artificial** é um sistema computacional formado por unidades de processamento interconectadas. Essas unidades, chamadas neurônios artificiais, organizam-se em **camadas** e podem aprender a resolver diferentes tarefas a partir de exemplos.

### Breve histórico

A seguir, estão alguns marcos históricos dos estudos sobre redes neurais artificiais:

- **Década de 1940**: McCulloch e Pitts propõem modelos lógicos inspirados em estruturas neurais.
- **Década de 1950**: pesquisadores apresentam novos modelos de redes neurais artificiais.
- **1959**: é desenvolvido o modelo adaptativo **ADALINE**.
- **1958**: Frank Rosenblatt apresenta o **Perceptron**.
- **1969**: Marvin Minsky e Seymour Papert publicam um livro sobre o Perceptron.
- **1982**: John Hopfield apresenta redes capazes de armazenar informações em suas conexões.
- **1984**: ocorre no Brasil um congresso dedicado a redes neurais.

### Redes neurais artificiais

Nos neurônios biológicos, sinais elétricos participam da transmissão de informações pelo sistema nervoso. Os modelos artificiais simplificam esse processo para representar computacionalmente alguns aspectos do aprendizado.

Uma unidade comum às redes neurais artificiais é o **neurônio artificial**. Ele recebe entradas, combina seus valores com pesos e produz uma saída. Os pesos representam a contribuição de cada entrada para o resultado.

O **ADALINE** (*Adaptive Linear Neuron*) usa uma combinação linear das entradas. O **Perceptron** aplica uma função de ativação à soma ponderada. Veja a representação de um perceptron:

![perceptron](Imagens/perceptron.png)

Na figura, $x_i$ representa as entradas, $w_i$ representa os pesos, e $g(.)$ representa a função de ativação. A saída do neurônio é representada por **y**.

Os pesos são ajustados durante o treinamento e influenciam o comportamento do neurônio. Em termos gerais:

- pesos positivos aumentam a contribuição de uma entrada;
- pesos negativos reduzem essa contribuição;
- os pesos definem como os sinais de entrada afetam a saída do modelo.

O treinamento busca valores de pesos adequados à tarefa.

O Perceptron de uma camada tem limitações para resolver problemas que não são linearmente separáveis. Uma extensão chamada **Perceptron multicamadas** (*Multi-Layer Perceptron*, MLP) usa várias camadas de neurônios.

- Pode ter múltiplas entradas, saídas e camadas.
- As entradas e saídas podem representar diferentes tipos de dados.
- Pode modelar relações que não são linearmente separáveis.

![multi-layer](Imagens/multi-layer.png)

Neste exemplo, o fluxo de informações é direto (*feedforward*): a saída de uma camada alimenta a camada seguinte. Uma rede pode conter desde poucos neurônios até muitas unidades organizadas em camadas.

Em uma rede multicamada, o fluxo costuma seguir esta ordem: **camada de entrada → camadas ocultas → camada de saída**. A saída pode ter um ou mais neurônios, conforme a tarefa e a forma de representar a previsão.

O número de camadas ocultas e de neurônios depende do problema. Não há uma quantidade única que funcione bem para todos os casos; esses valores são definidos por experimentação e validação.

- Quantidade de Neurônios = $\frac{Atributos Previsores + Atributos Meta}{2}$

Quando essa heurística é usada, o resultado deve ser arredondado para cima para obter um número inteiro de neurônios.

Cada camada oculta aplica operações aos valores recebidos, usando pesos e funções de ativação. A saída de uma camada torna-se a entrada da próxima. Durante o treinamento, os pesos são ajustados para reduzir o erro do modelo.

Durante o treinamento, o modelo calcula previsões para os exemplos e compara os resultados com os valores esperados. A partir do erro, ajusta os pesos e repete o processo até atingir um critério de parada.

Um Perceptron de uma camada só resolve problemas linearmente separáveis, nos quais é possível traçar uma fronteira linear entre as classes. Problemas que não atendem a essa condição exigem modelos mais expressivos.

### Como uma rede neural aprende?

Durante o treinamento, a rede ajusta os pesos para aproximar as saídas previstas dos resultados esperados. Ao final, os pesos armazenam os padrões aprendidos a partir dos exemplos.

Na prática, os **atributos preditores** são as entradas $x_i$. O modelo combina essas entradas com os pesos $w_i$ e aplica uma função de ativação para produzir a saída, que pode representar a classe prevista.
