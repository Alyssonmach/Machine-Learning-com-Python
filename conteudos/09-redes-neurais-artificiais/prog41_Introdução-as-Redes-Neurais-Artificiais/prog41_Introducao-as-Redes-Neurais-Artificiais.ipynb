# Introdução às redes neurais artificiais

Uma introdução aos conceitos e às aplicações das redes neurais.

### Introdução

A computação inclui algoritmos para resolver diferentes problemas, como sistemas de recomendação, buscas em GPS, construção de grafos e ordenação de dados. As redes neurais oferecem uma abordagem que pode aprender padrões a partir de exemplos e adaptar-se a diferentes tarefas.

As redes neurais são modelos computacionais inspirados, de forma simplificada, no funcionamento do cérebro. Elas são usadas em diversas aplicações de aprendizado de máquina, inclusive em métodos de aprendizado profundo (*deep learning*). O crescimento da capacidade de processamento e a disponibilidade de grandes volumes de dados contribuíram para ampliar seu uso.

### Aplicações

As redes neurais são usadas em diferentes áreas. Alguns exemplos incluem:

- Pesquisa e desenvolvimento de medicamentos;
- Processamento de linguagem natural;
- Veículos autônomos;
- Reconhecimento facial;
- Apoio à pesquisa médica;
- Análise de dados financeiros;
- Controle de tráfego;
- Geração de conteúdo artístico.

### Quando usar redes neurais?

Redes neurais podem ser úteis quando há muitos dados e o problema envolve relações complexas. Para problemas mais simples, outros métodos de aprendizado de máquina podem ser mais adequados.

### Em que as redes neurais se baseiam?

As redes neurais artificiais foram inspiradas no funcionamento de neurônios biológicos. Essa inspiração levou à criação de modelos computacionais capazes de aprender padrões a partir de exemplos.

Com o aumento da capacidade computacional e o desenvolvimento de novas técnicas, as redes neurais voltaram a ganhar popularidade e passaram a ser usadas em muitas aplicações.
