# Pesos e erros em redes neurais

Vamos estudar como os pesos são ajustados durante o treinamento de uma rede neural.

### Introdução

Nos materiais anteriores, vimos alguns tipos de redes neurais e como seus pesos influenciam as previsões. Agora, vamos observar como esses pesos podem ser ajustados.

Vamos analisar o cálculo dos pesos de cada neurônio.

### Inicialização dos pesos

Inicialmente, os pesos são definidos por um método de inicialização. Alguns métodos usam valores aleatórios; outros levam em conta as características dos dados para ajudar no treinamento.

Durante o treinamento, o modelo compara suas previsões com os valores esperados e calcula o erro. Em seguida, ajusta os pesos com base nesse erro e na **taxa de aprendizado**. Essa taxa controla o tamanho de cada ajuste: valores menores costumam produzir atualizações mais graduais. Uma forma simplificada de atualizar um peso é:

- $peso(n + 1) = peso(n) + TaxaAprendizagem * Entrada * Erro$  
- $erro = RespostaCorreta - RespostaCalculada$  

> "*O conhecimento de uma rede neural consiste em achar os pesos adequados que permitam o acerto dos resultados*".

### Funções de ativação

A função de ativação transforma a soma ponderada das entradas e determina a saída do neurônio. A escolha depende da arquitetura e do problema. A seguir, estão alguns exemplos:

- **Função degrau**: retorna valores discretos, como 0 ou 1.

![funcao-degrau](Imagens/funcao-degrau.png)

- **Função sigmoide**: retorna valores contínuos entre 0 e 1.

![funcao-sigmoid](Imagens/funcao-sigmoid.png)

- **Tangente hiperbólica**: retorna valores contínuos entre -1 e 1.

![funcao-tangente-hiperbolica](Imagens/funcao-tangente-hiperbolica.png)

A função de ativação deve ser escolhida de acordo com a arquitetura e a tarefa. Há várias opções, cada uma com propriedades e usos diferentes.

Em muitas arquiteturas, a função ReLU é usada nas camadas ocultas. A função de saída depende da tarefa: por exemplo, pode-se usar sigmoide para classificação binária e softmax para classificação multiclasse.

Dois termos relacionados às funções de ativação são:

- **Neurônio sigmoide**: neurônio cuja função de ativação é a sigmoide.
- **Perceptron**: modelo de neurônio que aplica uma função de ativação à combinação ponderada de suas entradas.

### Etapas simplificadas do treinamento

1. O modelo calcula previsões para os dados de treinamento.
2. Compara as previsões com os valores esperados e calcula o erro.
3. Atualiza os pesos e repete o processo até atingir um critério de parada.

![esquema-pesos-erros](Imagens/esquema-pesos-erros.png)

### Ajuste dos pesos

Após a inicialização, os pesos são atualizados durante o treinamento. Algumas técnicas e conceitos usados nesse processo são:

- Descida do gradiente (*gradient descent*);
- Derivadas;
- Regra delta;
- Retropropagação (*backpropagation*);
- Taxa de aprendizado (*learning rate*);
- Momento (*momentum*).

Algumas dessas técnicas serão discutidas com mais detalhes adiante.

### Medidas de erro

Duas medidas comuns para quantificar o erro de regressão são o **erro quadrático médio** (*Mean Squared Error*, MSE) e a **raiz do erro quadrático médio** (*Root Mean Squared Error*, RMSE):

- **MSE** = $\frac{1}{N}\sum_{i=1}^{n}(f_{i}-y_{i})^{2}$  
> $f_{i}$ = classe original e $y_{i}$ = classe obtida.  
- **RMSE** = $\sqrt{\frac{1}{N}\sum_{i=1}^{n}(f_{i}-y_{i})^{2}}{}$

O MSE e o RMSE resumem as diferenças entre os valores previstos e os valores observados. Como elevam os erros ao quadrado, penalizam mais fortemente as diferenças maiores.

### Termo de bias

O **bias** (termo de interceptação) é um parâmetro adicional somado à combinação ponderada das entradas. Ele permite que o modelo produza uma saída diferente de zero mesmo quando todas as entradas são zero.
