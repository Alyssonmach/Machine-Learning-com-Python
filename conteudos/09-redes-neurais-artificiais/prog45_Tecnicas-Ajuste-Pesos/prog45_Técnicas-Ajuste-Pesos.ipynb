# Técnicas para ajustar pesos

Vamos conhecer métodos usados para ajustar os pesos de uma rede neural.

### Introdução

Nos materiais anteriores, vimos que uma rede neural aprende ajustando seus pesos. Neste notebook, vamos estudar algumas técnicas usadas nesse processo.

### Descida do gradiente

O objetivo da descida do gradiente é reduzir a função de custo, que mede o erro do modelo. A cada etapa, o método ajusta os pesos na direção que tende a diminuir esse erro. Em problemas complexos, a função de custo pode ter vários mínimos locais, o que dificulta encontrar o mínimo global.

![descida-gradiente](Imagens/descida-gradiente.png)

Uma forma simplificada de representar o objetivo é minimizar a função de custo em relação aos pesos:

- $min_c(w_{1}, w_{2}, ..., w_{n})$, sendo c a função de custo (**cost function**);

Assim, a descida do gradiente busca valores de pesos que reduzam a função de custo.

Para escolher a direção do ajuste, calculamos as derivadas parciais da função de custo em relação aos pesos.

O **delta** representa um erro usado para ajustar os pesos das camadas de saída e das camadas ocultas.

A seguir, veja um esquema das etapas da descida do gradiente:

![metodo-gradiente](Imagens/metodo-gradiente.png)

Em uma superfície **convexa**, há um único mínimo global. A descida do gradiente em lote (*batch gradient descent*) calcula o erro usando todos os registros antes de atualizar os pesos.

Em uma superfície **não convexa**, podem existir mínimos locais. A descida do gradiente estocástica (*stochastic gradient descent*) atualiza os pesos após cada registro. A descida em minilotes (*mini-batch gradient descent*) usa um subconjunto de registros por atualização; o tamanho do lote pode ser definido pelo parâmetro `batch_size`.

Há diferentes variantes da descida do gradiente, que podem ser escolhidas de acordo com o problema:

- **Lote (*batch*)**: calcula o erro com todos os registros antes de atualizar os pesos.
- **Estocástica (*stochastic*)**: atualiza os pesos a cada registro.
- **Minilotes (*mini-batch*)**: atualiza os pesos após processar um grupo de registros, definido pelo parâmetro `batch_size`.

### Retropropagação (*backpropagation*)

A retropropagação calcula como o erro da saída depende dos pesos e usa essa informação para ajustá-los durante o treinamento.

Uma forma simplificada de representar a atualização de um peso é:

- $peso_{n+1} = (peso_{n} * momento) + (entrada * delta * TaxaAprendizagem)$;

Veja o significado de alguns parâmetros:

- **Taxa de aprendizado (*learning rate*)**: controla o tamanho de cada atualização dos pesos. Uma taxa alta pode acelerar o treinamento, mas também pode dificultar a convergência. Uma taxa baixa pode tornar o treinamento mais estável, porém mais lento.
- **Momento (*momentum*)**: usa parte das atualizações anteriores para orientar a próxima atualização e pode ajudar o treinamento a avançar com mais estabilidade.

### Outros métodos famosos

Há outras técnicas de otimização, como busca exaustiva, recozimento simulado (*simulated annealing*) e algoritmos genéticos. Cada técnica tem características próprias; a escolha depende do problema e do modelo.
