# Neurônio biológico

Uma visão simplificada da estrutura de um neurônio biológico.

As redes neurais artificiais foram inspiradas no funcionamento das redes neurais biológicas. Antes de estudar os modelos artificiais, vamos observar a estrutura básica de um neurônio.

> *Este material apresenta apenas uma visão introdutória do neurônio biológico, suficiente para acompanhar os próximos tópicos.*

### Estrutura de um neurônio biológico

![neuronio-biologico](Imagens/neuronio-biologico.jpg)

- **Neurônios**: células que processam e transmitem informações no sistema nervoso.
- **Axônio**: conduz sinais do neurônio até outras células.
- **Substâncias químicas**: participam da transmissão de sinais nas sinapses e podem alterar a atividade de outras células.
- **Sinapses**: conexões por meio das quais os neurônios transmitem sinais.

- Um neurônio pode ser ativado quando os sinais recebidos atingem determinado limiar.
- A atividade de uma rede de neurônios contribui para o processamento de informações e a produção de respostas.
