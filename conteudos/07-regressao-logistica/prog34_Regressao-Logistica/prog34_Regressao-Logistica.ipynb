# Regressão logística

Vamos estudar como usar a regressão logística para classificação.

#### Teoria sobre regressão logística

Apesar do nome, a **regressão logística** é um algoritmo de classificação, não de previsão de valores numéricos. A confusão costuma ocorrer por causa da semelhança com a **regressão linear**, que estima valores numéricos. A regressão logística é usada quando queremos classificar exemplos em categorias.

Enquanto a regressão linear ajusta uma função aos dados para estimar valores, a regressão logística usa a **função sigmoide** para estimar probabilidades entre 0 e 1. Essas probabilidades podem ser convertidas em classes usando um limiar definido para o problema.

![funcao-logistica.png](Imagens/funcao-logistica.png)

A função sigmoide recebe esse nome por causa de seu formato em S. Ela é definida por:

f(x) = $$\frac{1}{1 + e^{-x}}$$ para todo **x** real.  

A função sigmoide também é solução da seguinte equação diferencial:

 $\frac{dy}{dx}$ = &omega;(y)(1-y) com **y** entre 0 e 1.  

A função retorna valores entre 0 e 1. Quando `f(x)` é alto, o resultado se aproxima de 1; quando é baixo, aproxima-se de 0. Essa propriedade permite interpretar a saída como uma probabilidade e usá-la na classificação.

#### Como o algoritmo funciona

Durante o treinamento, o algoritmo estima os parâmetros que melhor se ajustam aos dados.

O algoritmo estima parâmetros para cada atributo, representados por $B_{0}, B_{1}, B_{2},...,B_{N}$. Esses parâmetros definem o peso de cada atributo. Um método comum para estimá-los é a **descida do gradiente**: o algoritmo testa diferentes valores e ajusta os parâmetros para reduzir o erro entre as previsões e os resultados observados.

![descida-gradiente.png](Imagens/descida-gradiente.png)

Com os parâmetros estimados, o algoritmo calcula uma combinação dos atributos de entrada e usa esse resultado como valor de **x** na função sigmoide. A probabilidade calculada pode então ser comparada a um limiar para escolher uma classe. Por exemplo, um limiar de 0,5 separa as classes conforme a probabilidade fique abaixo ou acima desse valor. A interpretação das classes depende de como os dados foram codificados no pré-processamento.
