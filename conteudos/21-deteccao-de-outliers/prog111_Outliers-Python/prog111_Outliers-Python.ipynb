# Detecção de outliers

Uma introdução à identificação de observações que se afastam do padrão dos dados.

![outlier](Imagens/outlier.gif)

**Outliers** são observações que diferem significativamente do padrão geral dos dados. Eles podem surgir por variações naturais, erros de registro ou fraudes.

O tratamento de outliers depende do problema e da origem dos dados. Podemos, por exemplo, manter as observações, removê-las ou substituir seus valores. Antes de tomar uma decisão, é importante considerar o contexto: os dados podem representar pessoas, eventos ou outros elementos reais.

O **boxplot** ajuda a visualizar a distribuição dos dados e possíveis valores atípicos. Ele resume os quartis, a mediana e os valores que ficam além dos limites definidos pelo método.
