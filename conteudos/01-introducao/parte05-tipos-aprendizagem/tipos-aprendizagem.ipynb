- **Aprendizado supervisionado**: durante o treinamento, os dados incluem os resultados esperados. O algoritmo aprende a relacionar os atributos de entrada às classes ou aos valores conhecidos. Depois, o modelo pode fazer previsões para novos dados.

- **Aprendizado não supervisionado**: o algoritmo analisa os dados sem receber rótulos ou respostas esperadas. Pode identificar associações e grupos; os padrões encontrados precisam ser interpretados no contexto do problema.

- **Aprendizado por reforço**: a máquina aprende por meio da interação com um ambiente e das consequências de suas ações. Por exemplo, um robô pode aprender a se deslocar recebendo recompensas ou penalidades pelas ações realizadas.
