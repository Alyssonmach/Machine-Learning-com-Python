# Termos Importantes

A seguir, apresentamos alguns termos importantes para compreender a inteligência artificial e as áreas relacionadas a ela.

> **Inteligência artificial**: área da Ciência da Computação dedicada ao desenvolvimento de sistemas capazes de realizar tarefas associadas à resolução de problemas.

> **Inteligência computacional**: termo usado por alguns pesquisadores em vez de inteligência artificial. Abrange estudos sobre redes neurais, computação evolucionária e lógica nebulosa (*fuzzy*).

> **Sistemas especialistas**: sistemas computacionais que representam conhecimentos de especialistas sobre determinados temas. Eles podem analisar informações e produzir respostas com base nesse conhecimento.

> **Visão computacional**: área que desenvolve sistemas capazes de interpretar imagens, por exemplo, para identificar objetos, pessoas e expressões faciais. Filtros de fotos e o desbloqueio de celulares por reconhecimento facial são algumas aplicações.

> **Aprendizado de máquina (*machine learning*)**: conjunto de métodos que permite a sistemas aprender padrões a partir de dados e usar esses padrões para realizar tarefas.

> **Processamento de linguagem natural**: área que estuda como fazer computadores processarem e interpretarem a linguagem humana.

> **Algoritmo genético**: método de busca e otimização inspirado nos princípios da seleção natural, proposta por Charles Darwin. Ele explora diferentes soluções para encontrar uma resposta aproximada para um problema.

> **Sistemas multiagentes**: sistemas formados por dois ou mais agentes que interagem ou cooperam para alcançar determinados objetivos.

> **Mineração de dados (*data mining*)**: processo de explorar grandes volumes de dados para encontrar padrões, por meio de métodos estatísticos e computacionais.

> **Lógica fuzzy (ou nebulosa)**: tipo de lógica multivalorada em que os valores podem variar entre 0 e 1. Diferentemente da lógica booleana, os valores não se limitam a 0 ou 1.

> **Raciocínio baseado em casos**: método que resolve novos problemas adaptando soluções usadas em problemas anteriores.

> **Redes neurais**: modelos de aprendizado de máquina compostos por unidades artificiais interconectadas, inspiradas de forma simplificada no funcionamento do cérebro.

> **Robótica**: área da tecnologia que desenvolve sistemas com componentes mecânicos e eletrônicos. Embora seja independente da inteligência artificial, a robótica pode se beneficiar da combinação com métodos de IA.

> **Aprendizado profundo (*deep learning*)**: área do aprendizado de máquina que usa redes neurais com várias camadas para aprender representações a partir de dados. O treinamento pode exigir grandes volumes de dados e maior capacidade computacional.

> **Big data**: termo usado para descrever grandes volumes de dados, que podem ser analisados para desenvolver modelos e sistemas computacionais.

> **Ciência de dados**: área interdisciplinar que combina conhecimentos de computação e estatística para explorar, analisar e interpretar dados.
