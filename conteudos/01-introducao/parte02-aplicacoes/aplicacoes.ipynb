# Aplicações de Machine Learning

O **aprendizado de máquina** (*machine learning*) busca desenvolver sistemas capazes de realizar tarefas que exigem algum tipo de raciocínio. Como a computação pode auxiliar o trabalho humano, sistemas com essas capacidades já fazem parte do nosso cotidiano. O aumento da capacidade de processamento e armazenamento, junto à disponibilidade de grandes volumes de dados (**big data**), ampliou as aplicações da inteligência artificial nos últimos anos.

A seguir, veja alguns exemplos de aplicações de aprendizado de máquina:

> 1. Reconhecimento de rostos em imagens;
> 2. Jogos que usam o Kinect, da Xbox;
> 3. Inteligência artificial em jogos para criar oponentes desafiadores;
> 4. Carros autônomos;
> 5. Robôs desenvolvidos para diferentes finalidades;
> 6. Aplicações na área da saúde;
> 7. Sistemas de recomendação baseados nas preferências dos usuários;
> 8. Apoio a vendas por meio da análise das preferências dos consumidores;
> 9. Aplicações desenvolvidas pela NASA para a exploração espacial;
> 10. Reconhecimento de voz e transcrição de fala;
> 11. Óculos de realidade aumentada;
> 12. Sistemas de busca, como o Google;
> 13. Pesquisa sobre capacidades humanas, como criatividade e expressão artística;
> 14. Detecção e prevenção de malware;
> 15. Redução de perdas e fraudes no fornecimento de energia elétrica;
> 16. Processamento de imagens;
> 17. Apoio à detecção precoce de doenças;
> 18. Automação de tarefas extensas, que podem ser executadas em menos tempo;
> 19. Serviços de atendimento ao cliente.

# Vale a pena aprender Machine Learning hoje?

Os estudos sobre inteligência artificial começaram há décadas. Hoje, porém, muitas empresas investem no desenvolvimento de tecnologias baseadas em IA, e suas aplicações alcançam diferentes áreas da vida cotidiana.

A redução de custos e o avanço das tecnologias também contribuíram para ampliar o uso da IA. Além disso, o volume de dados disponíveis cresce rapidamente. A análise desses dados permite desenvolver e aprimorar sistemas de inteligência artificial.

A popularização da internet e dos computadores pessoais, a partir dos anos 2000, transformou a tecnologia em poucas décadas. Esse histórico ajuda a entender por que o desenvolvimento tecnológico continua promissor. A inteligência artificial ocupa um papel importante nesse cenário. Como discute **Yuval Harari** em *Homo Deus: uma breve história do amanhã*, o conhecimento e a capacidade de interpretar informações influenciam a forma como o mundo se organiza. Por isso, é útil que profissionais de tecnologia conheçam os fundamentos de *machine learning*.
