# Método de classificação

A classificação é um método preditivo bastante usado em inteligência artificial. Neste material, vamos conhecer melhor seu funcionamento.

Com base nos **atributos preditores**, o modelo prevê uma **classe-alvo**. Em geral, os atributos preditores são variáveis de entrada, e a classe-alvo é a variável que queremos estimar. Por exemplo, podemos usar o histórico financeiro de uma pessoa para prever se conceder crédito a ela representa risco alto ou baixo. Nesse caso, os dados financeiros são os atributos preditores, e a categoria de risco é a classe-alvo.

Para treinar um algoritmo que classifique o risco de crédito de novos clientes, precisamos de uma base de dados com exemplos anteriores e suas classificações conhecidas. Com esses dados, o algoritmo aprende um **classificador**, capaz de receber informações de um novo cliente e atribuir-lhe uma classe.
