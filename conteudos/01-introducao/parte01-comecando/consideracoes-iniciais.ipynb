# Considerações iniciais sobre inteligência artificial

Antes de falarmos sobre inteligência artificial, vamos consultar algumas definições de **inteligência** em dicionários:

As definições mostram que não há uma explicação única e concisa para o termo *inteligência*. Por esse motivo, também não existe uma definição universal de inteligência artificial. Ainda assim, podemos compreender seus objetivos e métodos.

**Inteligência Artificial (IA)**, ou *Artificial Intelligence (AI)*, busca desenvolver sistemas capazes de pensar ou agir como seres humanos, ou de pensar e agir segundo princípios lógicos. Os avanços da computação tornaram os algoritmos cada vez mais refinados. Hoje, alguns sistemas conseguem interagir e analisar dados de maneira tão natural que podem parecer operados por pessoas. A partir de dados e treinamento, também podem fazer previsões e classificar elementos. Realizar tarefas semelhantes manualmente ou com programação tradicional seria trabalhoso. Por isso, a inteligência artificial se tornou útil para empresas de diversos setores. Embora pareça recente, a área tem uma longa história, que ganhou impulso durante a Segunda Guerra Mundial.

### O teste proposto por Alan Turing

Como vimos, não é possível definir inteligência artificial de forma totalmente precisa. Diante disso, durante a Segunda Guerra Mundial, o pesquisador **Alan Turing** propôs um teste para avaliar se uma máquina poderia demonstrar inteligência. A proposta se baseia no seguinte argumento:

No teste, uma pessoa conversa por texto com outro participante sem saber se ele é uma máquina ou um ser humano. Se, após algum tempo, a pessoa não conseguir distinguir os dois, considera-se que a máquina passou no teste de Turing.

Programar um computador para atender aos critérios propostos por Turing não é simples. A máquina precisaria processar linguagem natural, acessar uma ampla base de conhecimento — algo facilitado pela era do **big data** — e realizar raciocínio automatizado por meio de treinamento. Também precisaria aprender e se adaptar a novas circunstâncias, em vez de apenas repetir padrões.

### Fatos históricos relevantes para os estudos em inteligência artificial

- **(1943-1950)**: As pesquisas eram voltadas para o desenvolvimento de neurônios artificiais (**Redes Neurais**) para possibilitar o aprendizado de máquina (**Machine Learning**). Principais pesquisadores: **McCulloch & Pitts**.

- **1951–1969**: as máquinas, antes vistas principalmente como calculadoras, começaram a apresentar resultados em inteligência artificial. Algumas já eram capazes de jogar xadrez, provar teoremas lógicos e simular raciocínios humanos. Entre os principais pesquisadores estavam **Shannon**, **Turing**, **Newell**, **Simon**, **Weizenbaum**, **Evans** e **Buchanan**.

- **1970–1980**: pesquisadores da área se depararam com grandes limitações de armazenamento e de capacidade de processamento. Essas dificuldades desanimaram muitos estudiosos e reduziram os investimentos na área.

- **De 1981 até hoje**: com os avanços do hardware, as pesquisas em inteligência artificial voltaram a ganhar destaque. Com a expansão das tecnologias digitais, suas aplicações se tornaram mais práticas e específicas, alcançando setores como indústria, robótica, startups, mercados financeiros e marketing.

### Diferentes formas de pensar a inteligência artificial

- **Conexionista**: essa abordagem parte da ideia de que um modelo preciso do cérebro humano poderia permitir que uma máquina reproduzisse aspectos do conhecimento humano. Até 2020, os estudos de neurociência sobre a consciência — um dos fatores associados ao pensamento humano — ainda eram imprecisos. Isso mostra que ainda há muito a aprender sobre inteligência artificial.

- **Simbólica**: baseia-se na ideia de que estruturas simbólicas, como conjuntos de regras que manipulam dados, podem permitir a produção de comportamento inteligente em uma máquina.

- **Evolucionária**: inspirada nas teorias de **Charles Darwin**, essa abordagem usa seleção natural e eventos aleatórios para ajudar uma máquina a encontrar soluções. É uma forma de otimização também conhecida como **algoritmo genético**.

Combinar essas abordagens pode resultar em sistemas mais poderosos de aprendizado de máquina. A inteligência artificial abrange muitas áreas do conhecimento, cada uma com métodos próprios e desafios complexos.
