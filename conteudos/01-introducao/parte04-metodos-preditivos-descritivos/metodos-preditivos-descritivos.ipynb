# Métodos Preditivos e Descritivos

### Método Preditivo

O método preditivo busca estimar um resultado futuro com base em um conjunto de atributos.

- **Classificação**: atribui uma classe a cada registro. Por exemplo, pode prever se uma pessoa comprará determinado produto, se um pedido de crédito apresenta risco, se uma mensagem é spam ou a qual categoria uma notícia pertence. Também pode ser usado em tarefas de reconhecimento de voz e de rostos.

- **Regressão**: estima valores numéricos, em vez de classes. Por exemplo, pode prever vendas com base nos gastos com propaganda, estimar a velocidade do vento a partir de dados meteorológicos ou prever o preço de um imóvel com base em suas características.

### Método Descritivo

O método descritivo busca caracterizar os dados e identificar estruturas ou padrões neles.

- **Associação**: identifica itens ou eventos que ocorrem em conjunto. Por exemplo, pode ajudar a organizar produtos nas prateleiras, planejar promoções ou identificar fatores associados à evasão universitária.

- **Agrupamento**: organiza dados em grupos com características semelhantes. Pode ser usado para segmentar mercados, agrupar documentos e notícias ou identificar usuários influentes em redes sociais.

- **Detecção de desvios (*outliers*)**: identifica registros que se afastam de um padrão. Pode ajudar a detectar fraudes em cartões de crédito, intrusões em redes, padrões atípicos de consumo ou alterações no desempenho de máquinas.

- **Descoberta de padrões sequenciais**: identifica sequências de eventos que tendem a ocorrer em determinada ordem. Por exemplo, pode revelar padrões de compra e apoiar ações de marketing direcionadas.

- **Sumarização**: resume conjuntos de dados preservando as informações principais. Por exemplo, pode produzir um perfil conciso de um usuário a partir dos dados armazenados por uma empresa.
