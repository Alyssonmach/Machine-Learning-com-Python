# Máquina de vetores de suporte

Vamos estudar o funcionamento do SVM, um algoritmo usado em tarefas de classificação.

#### Introdução

O SVM é aplicado a diferentes tarefas, como reconhecimento de caracteres, voz e imagens. Seu desempenho depende dos dados, dos parâmetros e do problema. Redes neurais profundas também são usadas nessas aplicações.

#### Como funciona o SVM?

Em uma tarefa de classificação, o SVM busca encontrar um hiperplano que separe as classes com a maior margem possível. Em duas dimensões, esse hiperplano corresponde a uma reta. Os exemplos mais próximos da fronteira são chamados de **vetores de suporte**.

![svm-exemplo1](Imagens/svm-exemplo1.png)

Os vetores de suporte são os exemplos mais próximos da fronteira de decisão. Eles ajudam a definir a posição do hiperplano e a largura da margem entre as classes.

#### Definição formal

> "*Uma máquina de vetores de suporte (SVM, do inglês: support vector machine) é um conceito na ciência da computação para um conjunto de métodos de aprendizado supervisionado que analisam os dados e reconhecem padrões, usado para classificação e análise de regressão. O SVM padrão toma como entrada um conjunto de dados e prediz, para cada entrada dada, qual de duas possíveis classes a entrada faz parte, o que faz do SVM um classificador linear binário não probabilístico. Dados um conjunto de exemplos de treinamento, cada um marcado como pertencente a uma de duas categorias, um algoritmo de treinamento do SVM constrói um modelo que atribui novos exemplos a uma categoria ou outra. Um modelo SVM é uma representação de exemplos como pontos no espaço, mapeados de maneira que os exemplos de cada categoria sejam divididos por um espaço claro que seja tão amplo quanto possível. Os novos exemplos são então mapeados no mesmo espaço e preditos como pertencentes a uma categoria baseados em qual o lado do espaço eles são colocados.*"    

O objetivo do SVM é encontrar uma fronteira de decisão — chamada **hiperplano** — que separe exemplos de classes distintas. A margem é a distância entre essa fronteira e os exemplos mais próximos de cada classe.

O treinamento busca uma fronteira que separe as classes e maximize a margem. Em bases que não podem ser separadas perfeitamente, a formulação do SVM permite equilibrar a largura da margem e os erros de classificação.

#### Como definir a margem

Uma forma geométrica de explicar a separação é considerar os limites das regiões ocupadas por cada classe. O SVM posiciona a fronteira de decisão para maximizar a distância até os exemplos mais próximos, que são os vetores de suporte.

#### SVMs não lineares

Quando uma reta não consegue separar as classes, o SVM pode usar uma função de **kernel** para modelar uma fronteira não linear. O kernel permite representar relações em um espaço de maior dimensão sem calcular explicitamente toda a transformação.

#### Vantagens do SVM

O SVM pode ser útil em diferentes tarefas de aprendizado de máquina. Algumas características são:

1. Pode funcionar bem em espaços com muitas dimensões.
2. Pode ser usado em tarefas de classificação e regressão.
3. Com kernels, pode representar fronteiras de decisão não lineares.
4. Em algumas situações, exige menos ajustes que uma rede neural.

#### Desvantagens do SVM

Alguns pontos que devem ser considerados ao usar o SVM:

- Pode ser necessário testar diferentes combinações de parâmetros.
- O treinamento pode ser lento em bases muito grandes.
- Nem sempre é fácil interpretar o modelo e explicar suas previsões.
