# Agrupamento hierárquico

Uma introdução ao agrupamento hierárquico e aos dendrogramas.

O agrupamento hierárquico organiza os dados em grupos aninhados e representa as junções em um **dendrograma**. Em uma abordagem aglomerativa, começamos com cada registro em seu próprio grupo e, passo a passo, unimos os grupos mais próximos.

![clustering-dendrogram](Imagens/clustering-dendrogram.png)

- O resultado pode ser representado como uma árvore.
- Na abordagem **aglomerativa**, cada registro começa em seu próprio grupo e grupos próximos são unidos.
- As junções podem ser visualizadas em um **dendrograma**.

![hierarch](Imagens/hierarch.gif)

Para escolher o número de grupos, podemos traçar um corte horizontal no dendrograma e contar quantos ramos verticais ele atravessa.

- Escolha uma altura de corte que atravesse os ramos sem passar por uma junção.
- Conte os ramos verticais atravessados pelo corte. Essa quantidade corresponde ao número de grupos resultante.
