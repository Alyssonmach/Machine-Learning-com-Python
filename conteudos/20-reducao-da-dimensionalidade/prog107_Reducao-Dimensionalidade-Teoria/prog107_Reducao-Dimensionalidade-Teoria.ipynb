# Redução da dimensionalidade

Uma introdução a métodos usados para reduzir o número de variáveis.

Bases com muitas variáveis podem conter informações redundantes ou pouco relevantes. A redução da dimensionalidade transforma os dados para representá-los com menos variáveis. Isso pode reduzir o custo computacional e facilitar a análise. A seguir, veremos alguns métodos usados para esse fim.

### PCA (*Principal Component Analysis*)

1. **Seleção de características** escolhe algumas variáveis existentes; **extração de características** cria novas variáveis a partir das originais.
2. O PCA é um método não supervisionado.
3. O PCA identifica direções de maior variação nos dados e projeta os exemplos nessas direções.
4. A partir de **m** variáveis, o PCA pode gerar até **p ≤ m** componentes principais.
5. O número de componentes mantidos pode ser definido pelo usuário.

### LDA (*Linear Discriminant Analysis*)

1. O LDA é um método supervisionado que usa os rótulos das classes.
2. Busca projeções que aumentem a separação entre as classes.
3. O número de componentes obtidos é limitado pelo número de classes e pelas dimensões dos dados.

### Kernel PCA

1. O PCA representa variações lineares nos dados.
2. O Kernel PCA usa uma função de kernel para modelar estruturas não lineares.
3. Os componentes são calculados a partir das relações entre os dados no espaço definido pelo kernel.
