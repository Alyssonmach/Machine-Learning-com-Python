# Regressão linear

Vamos estudar como funciona esse método de aprendizado de máquina.

![regressao-linear](Imagens/regressao-linear.jpg)

A regressão linear estima valores numéricos. Diferentemente dos métodos de classificação, ela não prevê uma classe para cada exemplo; em vez disso, estima o valor de uma variável de interesse.

A regressão linear modela relações aproximadamente lineares entre variáveis. Para relações mais complexas, pode ser necessário usar transformações ou outros modelos de regressão.

A relação entre variáveis numéricas pode ser representada por uma variável dependente **y** e uma ou mais variáveis explicativas **x**. Alguns exemplos são:

- Temperatura, umidade e pressão do ar (**x**) → velocidade do vento (**y**);
- Gastos e histórico de compras (**x**) → limite do cartão (**y**);
- Idade (**x**) → custo do plano de saúde (**y**).

Em uma regressão linear simples, o valor estimado de **y** é calculado pela equação da reta:

$y = b_{0} + b_{1} * x_{1}$

O modelo busca encontrar os parâmetros que melhor representam a relação entre os dados. Nessa equação, $b_0$ é o intercepto, e $b_1$ é o coeficiente associado a $x_1$.

Uma medida comum para avaliar o erro do modelo é o **erro quadrático médio** (*Mean Squared Error*, MSE):

$MSE = \frac{1}{n}\sum_{i = 1}^{n}({f_{i} - y_{i})^{2}}$

O MSE resume as diferenças entre os valores previstos e os valores observados. Quanto mais distantes os pontos estiverem da reta, maior tende a ser o erro. O desempenho do modelo deve ser avaliado em dados que não foram usados no treinamento.

### Como estimar os parâmetros

- **Equação normal (álgebra linear)**: pode ser prática em bases com poucas variáveis, mas a inversão de matrizes pode ser custosa.
- **Descida do gradiente**: método iterativo que pode ser útil em problemas com muitas variáveis.

### Regressão linear múltipla

Na regressão linear múltipla, o modelo usa várias variáveis explicativas. O valor estimado de **y** pode ser representado assim:

$y = b_{0} + b_{1} * x_{1} + b_{2} * x_{2} + ... + b_{n} * x_{n}$
