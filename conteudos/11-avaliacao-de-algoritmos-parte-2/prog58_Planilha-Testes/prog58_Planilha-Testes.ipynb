# Planilha de testes

Uma planilha pode ajudar a registrar e comparar os resultados dos experimentos.

### Usando planilhas

Ao desenvolver um modelo de aprendizado de máquina, é útil comparar diferentes algoritmos e configurações. Registre o desempenho de cada classificador em várias divisões dos dados e, ao final, compare os resultados. A planilha incluída neste diretório pode ajudar nessa organização. Ela permite registrar métricas obtidas com diferentes sementes aleatórias, que podem gerar divisões distintas entre os conjuntos de treinamento e teste.
