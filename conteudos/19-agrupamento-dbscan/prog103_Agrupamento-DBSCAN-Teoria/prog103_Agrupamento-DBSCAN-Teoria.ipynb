# Algoritmo DBSCAN

Uma introdução ao algoritmo de agrupamento baseado em densidade.

**DBSCAN** (*Density-Based Spatial Clustering of Applications with Noise*) é um algoritmo de agrupamento que identifica regiões densas e separa pontos isolados como ruído.

O DBSCAN não exige que o número de grupos seja definido previamente. Em vez disso, usa parâmetros como o raio de vizinhança (`eps`) e o número mínimo de pontos (`min_samples`). A escolha desses valores influencia os grupos encontrados.

O desempenho do DBSCAN depende dos dados e dos parâmetros. Em alguns conjuntos, ele pode identificar formas de grupos que outros métodos não capturam.

O parâmetro `eps` define a distância máxima para considerar pontos como vizinhos.

### Características do DBSCAN

- Pode identificar grupos com formatos não lineares.
- Pode marcar pontos isolados como ruído (*outliers*).
- Não depende da inicialização de centroides.
- A escolha dos parâmetros pode ser desafiadora.
- Pontos de fronteira podem ser associados a grupos diferentes conforme a ordem de processamento.
