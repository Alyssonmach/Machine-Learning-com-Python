# Aprendizagem por Regras

O aprendizado de máquina baseado em regras analisa dados para gerar regras lógicas. Quando recebe um novo registro, o algoritmo aplica essas regras e produz uma saída de acordo com os dados informados.

A ideia é representar explicitamente o conhecimento sobre o problema por meio de regras. Ao comparar essas regras com os fatos observados, o sistema pode inferir novos resultados.

#### Algoritmo OneR

Ao desenvolver um sistema de inteligência artificial, é útil começar por métodos simples e avançar para os mais complexos conforme a necessidade. Assim, podemos escolher uma abordagem adequada à base de dados e evitar complexidade desnecessária.

O **OneR** (abreviação de *One Rule*) é um algoritmo de classificação simples. Ele cria uma regra para cada atributo preditor e escolhe aquela que apresenta o menor erro total. Para isso, constrói uma tabela de frequências entre cada preditor e a classe-alvo. As regras geradas são simples e fáceis de interpretar.

![oner.png](Imagens/oner.png)

A imagem mostra o algoritmo OneR aplicado a uma base de dados de crédito. Ele escolhe o atributo que produz o menor número de erros; neste exemplo, seleciona `renda`.

#### Algoritmo PRISM


O **PRISM** também gera regras de classificação. Enquanto o OneR usa um atributo por regra, o PRISM pode combinar vários atributos. O desempenho de cada método depende dos dados e do problema; usar mais atributos não garante, por si só, maior precisão.

#### Características da aprendizagem baseada em regras

- As regras costumam ser fáceis de compreender.

- O modelo pode ocupar pouco espaço de armazenamento.

- A construção do modelo pode ser mais lenta.

- Em algumas aplicações, as regras podem ter desempenho inferior ao de árvores de decisão.

- O método é menos usado quando outros algoritmos oferecem resultados mais adequados ao problema.
