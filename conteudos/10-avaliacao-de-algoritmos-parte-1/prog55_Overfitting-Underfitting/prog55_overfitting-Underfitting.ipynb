# Underfitting e overfitting

Vamos conhecer dois problemas comuns ao ajustar modelos de aprendizado de máquina.

### Underfitting (subajuste)

O subajuste ocorre quando o modelo é simples demais para representar os padrões dos dados. Por isso, pode apresentar baixo desempenho tanto no treinamento quanto em dados novos.

![underfitting](Imagens/underfitting.png)

À esquerda, há uma fronteira de decisão que separa adequadamente as classes. À direita, o modelo linear não representa bem a estrutura dos dados, o que leva a erros de classificação.

### Overfitting (sobreajuste)

O sobreajuste ocorre quando o modelo se adapta demais aos dados de treinamento. Ele pode ter bom desempenho nesses dados, mas fazer previsões inconsistentes em exemplos novos ou ligeiramente diferentes.

![overfitting](Imagens/overfitting.png)

À esquerda, o modelo representa a tendência geral dos dados. À direita, a fronteira se ajusta excessivamente aos exemplos de treinamento e pode não generalizar bem.

### Características

#### Underfitting (subajuste)

- Desempenho ruim nos dados de treinamento.

#### Overfitting (sobreajuste)

- Desempenho bom nos dados de treinamento.
- Desempenho ruim nos dados de teste.
- Modelo excessivamente específico.
- Possível memorização dos exemplos de treinamento.
- Maior sensibilidade a novos registros.

Por isso, é importante escolher um modelo adequado ao problema. Modelos simples podem resolver problemas simples; problemas mais complexos podem exigir métodos mais expressivos. A avaliação deve considerar dados que não foram usados no treinamento.
