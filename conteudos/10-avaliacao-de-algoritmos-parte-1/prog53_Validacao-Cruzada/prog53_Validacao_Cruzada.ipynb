# Validação cruzada

Veja como separar os dados para treinar e avaliar um modelo.

A validação cruzada estima o desempenho de um modelo em dados não usados diretamente no treinamento. O conjunto de dados é dividido em partes; em cada rodada, algumas partes são usadas para treinar o modelo e outra para avaliá-lo. Ao final, resumimos os resultados das rodadas.

A seguir, veremos três estratégias comuns de validação:

### Método holdout

- No método holdout, dividimos os dados em conjuntos de treinamento e teste. A proporção é definida previamente, e a divisão costuma ser aleatória.

![holdout](Imagens/holdout.jpg)

### Método k-fold

- No método k-fold, dividimos os dados em **k** partes de tamanho semelhante. Em cada rodada, uma parte é usada para teste e as demais para treinamento. O processo se repete até que cada parte tenha sido usada para teste.

![k-fold](Imagens/k-fold.jpg)

### Método leave-one-out

- O método leave-one-out é um caso particular do k-fold em que **k** é igual ao número de registros. Em cada rodada, um registro é usado para teste e os demais para treinamento. O método pode ser custoso computacionalmente e costuma ser mais viável em bases pequenas.

![leave-one-out](Imagens/leave-one-out.jpg)
