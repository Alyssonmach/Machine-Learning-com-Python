# Matriz de confusão

Veja como usar essa ferramenta para avaliar os resultados de algoritmos de aprendizado de máquina.

Uma matriz de confusão resume os acertos e erros de um classificador para cada classe. Em geral, as linhas representam as classes previstas, e as colunas, as classes reais. Os valores da diagonal mostram as classificações corretas; os demais indicam quais classes foram confundidas.

![matriz-confusao](Imagens/matriz-confusao.png)

A métrica adequada depende do problema. Em aplicações de maior risco, como saúde e crédito, é importante avaliar cuidadosamente os diferentes tipos de erro. Uma acurácia abaixo de 50% pode indicar que o modelo está pior do que uma escolha aleatória em algumas tarefas, mas a comparação depende da distribuição das classes e da métrica usada.
